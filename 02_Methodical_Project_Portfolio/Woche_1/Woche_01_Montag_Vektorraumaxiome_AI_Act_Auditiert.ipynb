## Block 1 (06:00 – 08:45): Vektorraumaxiome

**Lernziel:** Sie können die acht Axiome eines Vektorraums benennen, auf Beispiele anwenden und in Python (NumPy) überprüfen.

### 1.1 Definition eines Vektorraums

Ein Vektorraum über dem Körper $\mathbb{R}$ (den reellen Zahlen) ist eine nichtleere Menge $V$, ausgestattet mit einer **Addition** $+: V \times V \to V$ und einer **Skalarmultiplikation** $\cdot: \mathbb{R} \times V \to V$. Diese Abbildungen garantieren die **Abgeschlossenheit** (Ergebnisse liegen zwingend wieder in $V$). Darauf aufbauend müssen die folgenden acht Axiome gelten.

---

**Axiom 1: Kommutativität der Addition**

$$u + v = v + u \qquad \forall u,v \in V$$

---

**Axiom 2: Assoziativität der Addition**

$$(u + v) + w = u + (v + w) \qquad \forall u,v,w \in V$$

---

**Axiom 3: Existenz des neutralen Elements (Nullvektor)**

$$\exists 0 \in V \; \forall v \in V : v + 0 = v$$

---

**Axiom 4: Existenz des inversen Elements**

$$\forall v \in V \; \exists (-v) \in V : v + (-v) = 0$$

---

**Axiom 5: Neutralität der skalaren Eins**

$$1 \cdot v = v \qquad \forall v \in V$$

---

**Axiom 6: Assoziativität der Skalarmultiplikation**

$$a \cdot (b \cdot v) = (ab) \cdot v \qquad \forall a,b \in \mathbb{R}, v \in V$$

---

**Axiom 7: Distributivität der Skalarmultiplikation bzgl. der Vektoraddition**

$$a \cdot (u + v) = a \cdot u + a \cdot v \qquad \forall a \in \mathbb{R}, u,v \in V$$

---

**Axiom 8: Distributivität der Skalaraddition bzgl. der Vektormultiplikation**

$$(a + b) \cdot v = a \cdot v + b \cdot v \qquad \forall a,b \in \mathbb{R}, v \in V$$

---

```python
import numpy as np

# Systematischer Belastungstest über einen randomisierten reellen Raum (Monte-Carlo-Ansatz)
np.random.seed(42)
dimensions = (10000, 2)  # 10.000 Vektoren im R^2

U = np.random.uniform(-1000.0, 1000.0, dimensions)
V = np.random.uniform(-1000.0, 1000.0, dimensions)
W = np.random.uniform(-1000.0, 1000.0, dimensions)
A_scal = np.random.uniform(-10.0, 10.0, 10000)[:, np.newaxis]
B_scal = np.random.uniform(-10.0, 10.0, 10000)[:, np.newaxis]
zero_matrix = np.zeros(dimensions)

# Vektorisierte Massenverifikation aller 8 Axiome über die gesamte Matrix
assert np.allclose(U + V, V + U), "Axiom 1 (Kommutativität) verletzt"
assert np.allclose((U + V) + W, U + (V + W)), "Axiom 2 (Assoziativität) verletzt"
assert np.allclose(U + zero_matrix, U), "Axiom 3 (Nullvektor) verletzt"
assert np.allclose(U + (-U), zero_matrix), "Axiom 4 (Inverses Element) verletzt"
assert np.allclose(1 * U, U), "Axiom 5 (Skalare Eins) verletzt"
assert np.allclose(A_scal * (B_scal * U), (A_scal * B_scal) * U), "Axiom 6 (Skalare Assoziativität) verletzt"
assert np.allclose(A_scal * (U + V), A_scal * U + A_scal * V), "Axiom 7 (Distributivität I) verletzt"
assert np.allclose((A_scal + B_scal) * U, A_scal * U + B_scal * U), "Axiom 8 (Distributivität II) verletzt"

print("Strukturelle Verifikation: 10.000 Vektorkombinationen im R^2 fehlerfrei validiert (Toleranzgrenzen nach SR 11-7 Standard: rtol=1e-05, atol=1e-08).")
```

## Block 2 (09:30 – 11:40): Einführung in NumPy‑Tensoren

**Lernziel:** Sie können 0D‑, 1D‑ und 2D‑Tensoren mit NumPy erzeugen, deren `shape` und `dtype` kontrollieren sowie elementare Reduktionsoperationen entlang spezifischer Achsen durchführen.

### 2.1 Tensor-Dimensionen und Datenstruktur-Mapping

In der Datenanalyse repräsentieren Tensoren konkrete Datenstrukturen:
* **0D (Skalar):** Einzelner Messwert (z.B. ein Temperaturwert).
* **1D (Vektor):** Eindimensionale Sequenz (z.B. eine Zeitreihe oder ein Feature-Vektor).
* **2D (Matrix):** Tabellarische Daten (Zeilen = Beobachtungen/Samples, Spalten = Features/Merkmale).

### 2.2 Quantitatives Risiko-Mapping (3D-Tensor)

In der univariaten/multivariaten Simulation entspricht die Struktur der Shape `(Szenarien, Zeitschritte, Risikofaktoren)`. Eine Dimension von beispielsweise `(1000, 250, 4)` repräsentiert hierbei 1.000 simulierte Pfade (Monte Carlo) über 250 Handelstage für 4 makroökonomische Zinsstrukturkurven.

```python
import numpy as np

# Instanziierung unter expliziter Angabe des Datentyps (dtype)
s = np.array(5, dtype=np.int32)       # 0D
v = np.array([1.0, 2.0, 3.0])         # 1D
M = np.array([[1, 2], [3, 4]])        # 2D

# Verifikation von Dimensionen (shape) und Typen (dtype)
assert s.shape == () and s.dtype == np.int32
assert v.shape == (3,) and v.dtype == np.float64
assert M.shape == (2, 2)

# Effiziente Initialisierungsfunktionen für Platzhalter
zeros = np.zeros((3, 3))
ones = np.ones((2, 4))
```

### 2.3 Achsenbasierte Reduktionsoperationen
Reduktionen (wie .sum() or .mean()) aggregieren Daten entlang einer definierten Achse (axis). Die resultierende Dimension verliert exakt die kollabierte Achse.

```python
A = np.array([[1, 2],
              [3, 4]])

# Globale Reduktion (kollabiert den gesamten Tensor zu einem Skalar)
assert A.sum() == 10

# Reduktion über Achse 0 (kollabiert Zeilen, summiert Spaltenwerte)
assert np.array_equal(A.sum(axis=0), [4, 6])

# Reduktion über Achse 1 (kollabiert Spalten, summiert Zeilenwerte)
assert np.array_equal(A.sum(axis=1), [3, 7])
```

### 2.4 Systematische Übung: 3D-Tensoren
Ein 3D-Tensor (axis=0, 1, 2) repräsentiert typischerweise Simulationspfade oder Daten-Batches.

Aufgabe: Erzeugen Sie einen 3D‑Tensor der Form (2,3,4) mit np.random.rand. Berechnen Sie die Summe entlang jeder Achse und verifizieren Sie maschinell die resultierenden Dimensionen.

```python
np.random.seed(42)

# Erzeugung: 2 Batches, 3 Zeilen, 4 Spalten
rand3d = np.random.rand(2, 3, 4)

# Analyse des Dimensionsabbaus
assert rand3d.sum(axis=0).shape == (3, 4), "Fehler: Batches nicht korrekt aggregiert"
assert rand3d.sum(axis=1).shape == (2, 4), "Fehler: Zeilen nicht korrekt aggregiert"
assert rand3d.sum(axis=2).shape == (2, 3), "Fehler: Spalten nicht korrekt aggregiert"

print("Alle 3D-Reduktionsdimensionen erfolgreich verifiziert.")
```
