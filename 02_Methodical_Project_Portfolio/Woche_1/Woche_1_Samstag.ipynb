### 1.2 Normberechnung (Hand & Code)
**Aufgabe:** Berechnen Sie die Normen für den Vektor $x = (-3, 0, 4)$ manuell und validieren Sie das Ergebnis algorithmisch mittels testgetriebener Entwicklung.



```python
import numpy as np

# Manuelle Vorberechnung für das Protokoll:
# L1-Norm:   |-3| + |0| + |4| = 7
# L2-Norm:   sqrt((-3)^2 + 0^2 + 4^2) = sqrt(9 + 0 + 16) = sqrt(25) = 5
# Linf-Norm: max(|-3|, |0|, |4|) = 4

x = np.array([-3.0, 0.0, 4.0])

# Maschinelle Verifikation der Handrechnung
assert np.isclose(np.linalg.norm(x, 1), 7.0), "L1-Berechnung fehlerhaft"
assert np.isclose(np.linalg.norm(x, 2), 5.0), "L2-Berechnung fehlerhaft"
assert np.isclose(np.linalg.norm(x, np.inf), 4.0), "L∞-Berechnung fehlerhaft"

print("Status: Manuelle Berechnungen erfolgreich durch Assertions verifiziert.")
```
